```python
import shutil
import os

# UŻYWAMY PEŁNYCH ŚCIEŻEK (ABSOLUTNYCH)
# Dzięki temu unikamy błędu "No such file or directory"
base_path = "/content/datasets"

folder_glowny = os.path.join(base_path, "Cue-Detection-3")   # Docelowy
folder_dodatkowy = os.path.join(base_path, "cue-detection-1") # Źródłowy

subsets = ['train', 'valid', 'test']
dirs = ['images', 'labels']

print(f"Kopiowanie z: {folder_dodatkowy}")
print(f"Do: {folder_glowny}")

for subset in subsets:
    for d in dirs:
        src_dir = os.path.join(folder_dodatkowy, subset, d)
        dst_dir = os.path.join(folder_glowny, subset, d)

        # Sprawdzamy czy folder źródłowy istnieje
        if os.path.exists(src_dir):
            # Upewniamy się, że folder docelowy istnieje (jak nie, to go tworzymy)
            os.makedirs(dst_dir, exist_ok=True)

            files = os.listdir(src_dir)
            count = 0
            for file in files:
                src_file = os.path.join(src_dir, file)
                dst_file = os.path.join(dst_dir, file)

                # Kopiowanie
                shutil.copy(src_file, dst_file)
                count += 1

            print(f"Skopiowano {count} plików do {dst_dir}")
        else:
            print(f"Pominięto (brak folderu źródłowego): {src_dir}")

print("Zakończono łączenie datasetów.")
```

## Train a YOLOv8 Keypoint Detection Model

With our dataset downloaded, we can now train a YOLOv8 keypoint detection model. Run the code snippet below to start training your model:

```python
%cd {HOME}
from ultralytics import YOLO

model = YOLO('yolov8n-pose.pt')

results = model.train(data=f"{dataset.location}/data.yaml", epochs=150, imgsz=640, patience=30, batch=16)
```

Your model will train for 100 epochs. After training, you can run test your model using an image from your test set.

## Test the Keypoint Detection Model

Let's test our keypoint detection model using [supervision](https://github.com/roboflow/supervision).

```python
%cd {HOME}
model = YOLO('runs/pose/train/weights/best.pt')

import os
import random

random_file = random.choice(os.listdir(f"{dataset.location}/test/images"))
file_name = os.path.join(f"{dataset.location}/test/images", random_file)

results = model(file_name)

print(results[0])
```

We can visualize our keypoints using the following code:

```python
!pip install supervision==0.24.0 -q

import supervision as sv
import cv2

key_points = sv.KeyPoints.from_ultralytics(results[0])


vertex_annotator = sv.VertexAnnotator(radius=3, color=sv.Color.WHITE)
edge_annotator = sv.EdgeAnnotator(thickness=2, edges=[(0, 1)])

annotated_frame = cv2.imread(file_name)
annotated_frame = edge_annotator.annotate(
    scene=annotated_frame,
    key_points=key_points)
annotated_frame = vertex_annotator.annotate(
    scene=annotated_frame,
    key_points=key_points)

annotated_frame = sv.resize_image(
    annotated_frame,
    resolution_wh=(900, 900),
    keep_aspect_ratio=True
)
sv.cv2_to_pillow(annotated_frame)
```

Our model successfully identified keypoints in the image.

```python
import matplotlib.pyplot as plt
import numpy as np
from ultralytics import YOLO
import torch

# --- KONFIGURACJA ŚCIEŻEK ---
# Upewnij się, że te ścieżki są poprawne w Twoim Colabie
model_path = '/content/cue-detection/runs/pose/train/weights/best.pt'
data_path = '/content/cue-detection/datasets/Cue-Detection-3/data.yaml'

# --- STYL (Times New Roman, Duże Czcionki) ---
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'font.size': 14,
    'axes.titlesize': 18,
    'axes.labelsize': 16,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'legend.fontsize': 12,
    'figure.dpi': 300,
    'lines.linewidth': 2
})

def generate_curves_with_classes():
    print("Ładowanie modelu i walidacja...")
    try:
        model = YOLO(model_path)
        metrics = model.val(data=data_path, split='test', plots=False)
    except Exception as e:
        print(f"BŁĄD: {e}")
        return

    # Pobieramy nazwy klas (np. {0: 'cue ball', 1: 'other'})
    names = model.names
    num_classes = len(names)

    # Oś X dla Confidence (0-1)
    x_conf = np.linspace(0, 1, 1000)

    try:
        # Pobieramy krzywe (kształt: [Liczba_Klas, 1000])
        # p_curve, r_curve, f1_curve zawierają dane dla każdej klasy osobno
        P_curves = metrics.box.p_curve
        R_curves = metrics.box.r_curve
        F1_curves = metrics.box.f1_curve

        # Obliczamy średnie (All Classes)
        P_all = P_curves.mean(0)
        R_all = R_curves.mean(0)
        F1_all = F1_curves.mean(0)

    except AttributeError:
        print("Nie udało się pobrać krzywych. Sprawdź wersję biblioteki.")
        return

    # --- WYKRES 1: F1-CONFIDENCE CURVE (z klasami) ---
    plt.figure(figsize=(10, 7))

    # Rysujemy linie dla poszczególnych klas (cieńsze)
    for i in range(num_classes):
        class_name = names[i]
        plt.plot(x_conf, F1_curves[i], linewidth=1.5, alpha=0.7, label=f'{class_name}')

    # Rysujemy linię średnią (gruba, niebieska)
    best_idx = np.argmax(F1_all)
    max_f1 = F1_all[best_idx]
    best_conf = x_conf[best_idx]

    plt.plot(x_conf, F1_all, color='blue', linewidth=3, label=f'all classes ({max_f1:.2f} at {best_conf:.2f})')

    # Kropka max F1
    plt.scatter(best_conf, max_f1, color='blue', s=80, zorder=5)

    plt.title('F1-Confidence Curve', fontweight='bold', pad=15)
    plt.xlabel('Confidence')
    plt.ylabel('F1')
    plt.xlim(0, 1)
    plt.ylim(0, 1.05)
    plt.legend(loc='lower left')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.savefig('F1_Curve_Classes.pdf', bbox_inches='tight')
    print("Zapisano: F1_Curve_Classes.pdf")

    # --- WYKRES 2: PRECISION-RECALL CURVE (z klasami) ---
    plt.figure(figsize=(10, 7))

    # Dla PR Curve osią X jest zazwyczaj Recall, a Y Precision
    # Musimy iterować po klasach i brać pary (R, P)

    for i in range(num_classes):
        class_name = names[i]
        # Wyliczamy mAP dla danej klasy (w przybliżeniu średnia Precision)
        ap_val = metrics.box.ap50[i]
        plt.plot(R_curves[i], P_curves[i], linewidth=1.5, alpha=0.7, label=f'{class_name} ({ap_val:.3f} mAP@0.5)')

    # Linia średnia (All Classes)
    map_val = metrics.box.map50
    plt.plot(R_all, P_all, color='blue', linewidth=3, label=f'all classes ({map_val:.3f} mAP@0.5)')

    plt.title('Precision-Recall Curve', fontweight='bold', pad=15)
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.xlim(0, 1)
    plt.ylim(0, 1.05)
    plt.legend(loc='lower left')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.savefig('PR_Curve_Classes.pdf', bbox_inches='tight')
    print("Zapisano: PR_Curve_Classes.pdf")

    # Możesz dodać analogicznie wykresy P-Confidence i R-Confidence jeśli potrzebujesz

if __name__ == "__main__":
    generate_curves_with_classes()
```

```python
import matplotlib.pyplot as plt
import matplotlib.font_manager as font_manager
import seaborn as sns
import numpy as np
import os

# --- 1. ŁADOWANIE CZCIONKI TIMES NEW ROMAN (DLA COLAB) ---
# Jeśli masz już czcionkę z poprzedniego kroku, ten blok po prostu zadziała.
if not os.path.exists('times.ttf'):
    !wget -q -O times.ttf "https://github.com/justrajdeep/fonts/blob/master/Times%20New%20Roman.ttf?raw=true"

try:
    font_manager.fontManager.addfont('times.ttf')
    plt.rcParams.update({
        'font.family': 'serif',
        'font.serif': ['Times New Roman'],
        'font.size': 16,
        'axes.titlesize': 24,
        'axes.labelsize': 20,
        'xtick.labelsize': 16,
        'ytick.labelsize': 16,
        'figure.dpi': 300
    })
    print("Czcionka Times New Roman załadowana.")
except Exception as e:
    print(f"Uwaga: Nie udało się załadować czcionki ({e}). Używam domyślnej.")

# --- 2. DANE MACIERZY (Przepisane z Twojego obrazka) ---
# Wiersze: True Label (Prawdziwa klasa)
# Kolumny: Predicted Label (Przewidziana klasa)
# Kolejność: [Cue Ball, Other, Background]

data = np.array([
    [70, 6],   # True: Cue Ball
    [4, 0],   # True: Other
])

# Etykiety klas (takie jak na osiach)
labels = ['Cue', 'Background']

# --- 3. RYSOWANIE ---
plt.figure(figsize=(10, 8))

# Generowanie heatmapy
# fmt='g' zapobiega notacji naukowej (np. 1e3)
# cmap='Blues' to standardowa, czytelna kolorystyka
sns.set_context("paper", rc={"font.family": "serif", "font.serif": ["Times New Roman"]}) # Wymuszenie czcionki w Seaborn
ax = sns.heatmap(data, annot=True, fmt='g', cmap='Blues',
                 xticklabels=labels, yticklabels=labels,
                 annot_kws={"size": 18, "weight": "bold"})

# Dostosowanie opisów osi
ax.set_title('Confusion Matrix', fontsize=26, fontweight='bold', pad=20)
ax.set_xlabel('Predicted', fontsize=20, labelpad=10)
ax.set_ylabel('True', fontsize=20, labelpad=10)

# Powiększenie etykiet na osiach (Cue Ball, Other...)
ax.set_xticklabels(labels, fontsize=18)
ax.set_yticklabels(labels, fontsize=18, rotation=90, va='center') # Pionowy tekst na osi Y

# Poprawki estetyczne paska legendy (colorbar)
cbar = ax.collections[0].colorbar
cbar.ax.tick_params(labelsize=14)

# --- 4. ZAPIS ---
plt.tight_layout()
plt.savefig('Confusion_Matrix_HighRes.pdf', bbox_inches='tight')
print("Zapisano: Confusion_Matrix_HighRes.pdf")
plt.show()
```
